# **Carga de datos**

```python
from google.colab import files
from pathlib import Path
import zipfile

subidos = files.upload()
carpeta = Path("/content/cloud_provider")
carpeta.mkdir(exist_ok=True)

for nombre in subidos:
    if nombre.lower().endswith(".zip"):
        with zipfile.ZipFile(nombre) as archivo_zip:
            archivo_zip.extractall(carpeta)
    elif nombre.lower().endswith(".csv"):
        destino = carpeta / nombre
        destino.write_bytes(Path(nombre).read_bytes())

print("Archivos CSV encontrados:")
for ruta in carpeta.rglob("*.csv"):
    print(ruta)
```

# **Datos crudos inmutables: los archivos de Landing no deben modificarse.**

```python
import pandas as pd

archivos_csv = list(carpeta.rglob("*.csv"))
tablas = {}

for ruta in archivos_csv:
    df = pd.read_csv(ruta, dtype="string", keep_default_na=False)
    df = df.replace(r"^\s*$", pd.NA, regex=True)
    tablas[ruta.name] = df

    print(f"\n--- {ruta.name} ---")
    print("Filas y columnas:", df.shape)
    display(df.head(5))
```

```python
import json
from collections import Counter

archivos_jsonl = sorted(carpeta.rglob("*.jsonl"))
print("Archivos JSONL encontrados:", len(archivos_jsonl))

registros = []
archivo_de_cada_evento = []

for ruta in archivos_jsonl:
    cantidad = 0

    with open(ruta, encoding="utf-8") as archivo:
        for linea in archivo:
            if linea.strip():
                registros.append(json.loads(linea))
                archivo_de_cada_evento.append(ruta.name)
                cantidad += 1

    print(f"{ruta.name}: {cantidad} eventos")

if registros:
    eventos = pd.DataFrame(registros)
    print("\nTotal de eventos:", len(eventos))
    display(eventos.head())
else:
    print("No hay eventos para analizar todavía.")
```

# **Exploracion de nulos**

```python
for nombre, df in tablas.items():
    nulos = df.isna().sum()
    nulos = nulos[nulos > 0]

    print(f"\n{nombre}")
    if nulos.empty:
        print("  Sin nulos")
    else:
        for columna, cantidad in nulos.items():
            print(f"  {columna}: {cantidad} nulos")
```

# **Valores nulos y atípicos: existen costos negativos ocasionales y spikes de magnitud elevada.**

```python
eventos_revision = eventos.copy()
eventos_revision["value_num"] = pd.to_numeric(
    eventos_revision["value"], errors="coerce"
)

no_convertibles = (
    eventos_revision["value"].notna()
    & eventos_revision["value_num"].isna()
)
print("value no convertibles:", int(no_convertibles.sum()))

for metrica, grupo in eventos_revision.groupby("metric", dropna=False):
    valores = grupo["value_num"].dropna()

    if valores.empty:
        continue

    q1 = valores.quantile(0.25)
    q3 = valores.quantile(0.75)
    limite = q3 + 1.5 * (q3 - q1)

    print(f"\n{metrica}")
    print("Eventos:", len(grupo))
    print("Valores informados:", len(valores))
    print("Mínimo:", valores.min())
    print("Mediana:", valores.median())
    print("Percentil 99:", valores.quantile(0.99))
    print("Máximo:", valores.max())
    print("Valores sobre el límite IQR:", int((valores > limite).sum()))
```

> Q1 es el valor por debajo del cual está el 25 % de los datos, y Q3, el valor por debajo del cual está el 75 %.

> Entonces el IQR mide el rango donde está el 50 % central de los valores.





```python
for metrica, grupo in eventos_revision.groupby("metric", dropna=False):
    valores = grupo["value_num"].dropna()

    print(
        f"{metrica}: "
        f"{int((valores < 0).sum())} negativos; "
        f"mínimo observado = {valores.min()}"
    )
```

```python
campos_no_negativos = [
    "cpu_hours",
    "requests",
    "storage_gb_hours",
    "carbon_kg",
    "genai_tokens"
]

for campo in campos_no_negativos:
    if campo not in eventos.columns:
        continue

    valores = pd.to_numeric(eventos[campo], errors="coerce").dropna()

    print(
        f"{campo}: "
        f"{int((valores < 0).sum())} negativos; "
        f"mínimo observado = {valores.min()}"
    )

if "cost_usd_increment" in eventos.columns:
    costos = pd.to_numeric(
        eventos["cost_usd_increment"], errors="coerce"
    ).dropna()

    print(
        f"\ncost_usd_increment: "
        f"{int((costos < 0).sum())} negativos; "
        f"mínimo observado = {costos.min()}"
    )
```



> Analisis de valores negativos segun el contexto del campo.


> Como se comenta en la consigna, el campo cost_usd_increment presenta valores negativos.





```python
costos = eventos.copy()
costos["cost_num"] = pd.to_numeric(
    costos["cost_usd_increment"], errors="coerce"
)

resultados = []

for servicio, grupo in costos.groupby("service", dropna=False):
    valores = grupo["cost_num"].dropna()

    q1 = valores.quantile(0.25)
    q3 = valores.quantile(0.75)
    limite_iqr = q3 + 1.5 * (q3 - q1)

    resultados.append({
        "servicio": servicio,
        "eventos": len(grupo),
        "costos_informados": len(valores),
        "negativos": int((valores < 0).sum()),
        "mediana": valores.median(),
        "p99": valores.quantile(0.99),
        "maximo": valores.max(),
        "limite_superior_iqr": limite_iqr,
        "posibles_spikes_iqr": int((valores > limite_iqr).sum())
    })

display(
    pd.DataFrame(resultados)
    .sort_values("posibles_spikes_iqr", ascending=False)
)
```

> Sí aparece evidencia clara de posibles spikes. En analytics por ejemplo, el máximo de cost_usd_increment es 249,38, frente a un P99 de 11,60 y un límite superior IQR de 20,10. Es decir, el máximo supera ampliamente ambas referencias.



```python
campos_csv = {
    "customers_orgs.csv": ["nps_score"],
    "nps_surveys.csv": ["nps_score"],
    "support_tickets.csv": ["csat"],
    "billing_monthly.csv": [
        "subtotal", "credits", "taxes", "exchange_rate_to_usd"
    ]
}

resultados = []

for archivo, columnas in campos_csv.items():
    df = tablas[archivo]

    if archivo == "billing_monthly.csv":
        grupos = list(df.groupby("currency", dropna=False))
    else:
        grupos = [("todos", df)]

    for grupo, datos in grupos:
        for columna in columnas:
            original = datos[columna]
            numero = pd.to_numeric(original, errors="coerce")
            valores = numero.dropna()

            if valores.empty:
                continue

            q1 = valores.quantile(0.25)
            q3 = valores.quantile(0.75)
            iqr = q3 - q1

            resultados.append({
                "archivo": archivo,
                "grupo": grupo,
                "columna": columna,
                "nulos": int(original.isna().sum()),
                "no_convertibles": int(
                    (original.notna() & numero.isna()).sum()
                ),
                "minimo": valores.min(),
                "mediana": valores.median(),
                "p99": valores.quantile(0.99),
                "maximo": valores.max(),
                "negativos": int((valores < 0).sum()),
                "debajo_limite_IQR": int(
                    (valores < q1 - 1.5 * iqr).sum()
                ),
                "encima_limite_IQR": int(
                    (valores > q3 + 1.5 * iqr).sum()
                )
            })

display(pd.DataFrame(resultados))
```



*   csat podria ir del 0-10 como encuesta, lo veo bien aunque el rango original es de 1-5.
*   El nps_score parece venir en un rango de -100 a 100 y no de 0-10. Sin embargo tenemos un peak de 101 en customer_orgs, esto es un outlier por contexto.
*   El subtotal parece estar en negativo en algunos casos, pero al ser facturacion podria ser un ajuste, no parece un inconveniente.
